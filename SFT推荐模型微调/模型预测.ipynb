```python
import json
from openai import OpenAI
from tqdm import tqdm

# 加载模型
client = OpenAI(
    api_key="EMPTY",
    # 需要修改为大模型地址
    base_url="http://xx.xx.xx.xx:8000/v1/"
)
```

```python
# 加载测试数据
test_file_path = "./data/processed/mind_test.json"
with open(test_file_path, "r", encoding='utf-8') as test_file:
    test_data = json.load(test_file)
print(len(test_data))
```

```python
# 开始预测
labels = []
predictions = []
for each_test in tqdm(test_data):
    chat_completion = client.chat.completions.create(
        messages=[
            {
                "role": "system",
                "content": each_test["instruction"]
            },
            {
                "role": "user",
                "content": each_test["input"],
            }
        ],
        model="glm4",
    )

    predictions.append(chat_completion.choices[0].message.content)
    labels.append(each_test["output"])

    if len(labels) % 10 == 0:
        correct = 0
        wrong = 0

        for l, p in zip(labels, predictions):
            l = l.strip()
            p = p.strip()
            if l == p:
                correct += 1
            else:
                wrong += 1
        print(f'总样本数：{len(labels)}，准确数：{correct}, 错误数：{wrong}, 准确率：{correct / len(labels)}')


assert len(predictions) == len(labels)

correct = 0
wrong = 0

for l, p in zip(labels, predictions):
    l = l.strip()
    p = p.strip()
    if l == p:
        correct += 1
    else:
        wrong += 1
print(f'总样本数：{len(labels)}，准确数：{correct}, 错误数：{wrong}, 准确率：{correct/len(labels)}')
```
